## Solutions: Part A (Functions)

**A1 · Even or odd**

````python
def is_even(n):
    """Return True if n is even."""
    return n % 2 == 0
````

**A2 · Default parameters**

````python
def full_name(first, last, middle=""):
    parts = [first, middle, last]
    return " ".join(p for p in parts if p)
````

**A3 · Temperature conversion**

````python
def celsius_to_fahrenheit(c):
    return c * 9 / 5 + 32
````

**A4 · Count vowels**

````python
def vowel_count(text):
    return sum(1 for ch in text.lower() if ch in "aeiou")
````

**A5 · Variable arguments with *args**

````python
def average(*numbers):
    if not numbers:
        return None
    return sum(numbers) / len(numbers)
````

**A6 · Keyword arguments with **kwargs**

````python
def build_profile(name, **details):
    profile = {"name": name}
    profile.update(details)
    return profile
````

**A7 · Returning multiple values**

````python
def min_max_avg(values):
    return min(values), max(values), sum(values) / len(values)
````

**A8 · Passing functions as arguments**

````python
def apply_twice(func, x):
    return func(func(x))
````

**A9 · Fix the mutable default bug**

````python
def add_student(name, roster=None):
    if roster is None:
        roster = []
    roster.append(name)
    return roster
````

**A10 · Closures: a configurable counter**

````python
def make_counter(start=0, step=1):
    current = start
    def counter():
        nonlocal current
        value = current
        current += step
        return value
    return counter
````

**A11 · Custom sort with lambda**

````python
def sort_words(words):
    return sorted(words, key=lambda w: (len(w), w))
````

**A12 · Decorator: count calls**

````python
def count_calls(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        wrapper.calls += 1
        return func(*args, **kwargs)
    wrapper.calls = 0
    return wrapper
````

**A13 · Decorator factory: retry**

````python
def retry(times):
    def decorator(func):
        @wraps(func)
        def wrapper(*args, **kwargs):
            last_error = None
            for _ in range(times):
                try:
                    return func(*args, **kwargs)
                except Exception as error:
                    last_error = error
            raise last_error
        return wrapper
    return decorator
````

**A14 · Function composition**

````python
def compose(*funcs):
    def composed(x):
        for f in reversed(funcs):
            x = f(x)
        return x
    return composed
````

**A15 · Positional-only and keyword-only parameters**

````python
def make_url(host, /, path="", *, secure=True):
    scheme = "https" if secure else "http"
    url = f"{scheme}://{host}"
    if path:
        url += f"/{path}"
    return url
````

## Solutions: Part B (Recursion)

**B1 · Countdown list**

````python
def count_down_list(n):
    if n <= 0:
        return []
    return [n] + count_down_list(n - 1)
````

**B2 · Sum of first n numbers**

````python
def sum_to_n(n):
    if n == 0:
        return 0
    return n + sum_to_n(n - 1)
````

**B3 · Power function**

````python
def power(base, exp):
    if exp == 0:
        return 1
    return base * power(base, exp - 1)
````

**B4 · Count digits**

````python
def count_digits(n):
    if n < 10:
        return 1
    return 1 + count_digits(n // 10)
````

**B5 · Reverse a list**

````python
def reverse_list(lst):
    if len(lst) <= 1:
        return lst[:]
    return reverse_list(lst[1:]) + [lst[0]]
````

**B6 · Greatest common divisor**

````python
def gcd(a, b):
    if b == 0:
        return a
    return gcd(b, a % b)
````

**B7 · Palindrome check**

````python
def is_palindrome(s):
    if len(s) <= 1:
        return True
    if s[0] != s[-1]:
        return False
    return is_palindrome(s[1:-1])
````

**B8 · Fibonacci with memoization**

````python
@lru_cache(maxsize=None)
def fib(n):
    if n < 2:
        return n
    return fib(n - 1) + fib(n - 2)
````

**B9 · Recursive binary search**

````python
def binary_search(arr, target, lo=0, hi=None):
    if hi is None:
        hi = len(arr) - 1
    if lo > hi:
        return -1
    mid = (lo + hi) // 2
    if arr[mid] == target:
        return mid
    if arr[mid] < target:
        return binary_search(arr, target, mid + 1, hi)
    return binary_search(arr, target, lo, mid - 1)
````

**B10 · Flatten a nested list**

````python
def flatten(items):
    flat = []
    for item in items:
        if isinstance(item, list):
            flat.extend(flatten(item))
        else:
            flat.append(item)
    return flat
````

**B11 · Sum of a nested list**

````python
def nested_sum(data):
    total = 0
    for item in data:
        if isinstance(item, list):
            total += nested_sum(item)
        else:
            total += item
    return total
````

**B12 · Power set (all subsets)**

````python
def power_set(items):
    if not items:
        return [[]]
    rest = power_set(items[1:])
    return rest + [[items[0]] + subset for subset in rest]
````

**B13 · Permutations**

````python
def permutations(items):
    if len(items) <= 1:
        return [list(items)]
    result = []
    for i in range(len(items)):
        rest = items[:i] + items[i + 1:]
        for perm in permutations(rest):
            result.append([items[i]] + perm)
    return result
````

**B14 · Merge sort**

````python
def merge_sort(arr):
    if len(arr) <= 1:
        return arr[:]
    mid = len(arr) // 2
    left, right = merge_sort(arr[:mid]), merge_sort(arr[mid:])
    merged, i, j = [], 0, 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            merged.append(left[i]); i += 1
        else:
            merged.append(right[j]); j += 1
    return merged + left[i:] + right[j:]
````

**B15 · Tower of Hanoi**

````python
def hanoi_moves(n, source="A", target="C", spare="B"):
    if n == 1:
        return [(source, target)]
    return (hanoi_moves(n - 1, source, spare, target)
            + [(source, target)]
            + hanoi_moves(n - 1, spare, target, source))
````

**B16 · Generate balanced parentheses**

````python
def generate_parentheses(n):
    result = []
    def build(current, opened, closed):
        if len(current) == 2 * n:
            result.append(current)
            return
        if opened < n:
            build(current + "(", opened + 1, closed)
        if closed < opened:
            build(current + ")", opened, closed + 1)
    build("", 0, 0)
    return result
````

**B17 · Subset sum (backtracking)**

````python
def subset_sum(nums, target):
    if target == 0:
        return True
    if not nums:
        return False
    first, rest = nums[0], nums[1:]
    return subset_sum(rest, target - first) or subset_sum(rest, target)
````

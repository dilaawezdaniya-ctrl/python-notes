## 1. What is a Loop?

A **loop** repeats a block of code multiple times, so you don't have to write the same instructions again and again.

Python has two loop statements:

| Loop | Use when... |
|------|-------------|
| `for` | You know *what* you are iterating over (a collection or a fixed count) — **definite iteration** |
| `while` | You repeat *until a condition becomes false* — **indefinite iteration** |

**Without a loop:**

````python
print("Hello 1")
print("Hello 2")
print("Hello 3")
````

**With a loop:**

````python
for i in range(1, 4):
    print("Hello", i)
````

## 2. The `for` Loop

### Syntax
```python
for variable in iterable:
    # loop body (indented)
```

Python's `for` loop takes each item from an *iterable* (list, tuple, string, dict, set, range, file, ...) one at a time.

### 2.1 Iterating over a list

````python
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
````

### 2.2 Iterating over a string

````python
for ch in "Python":
    print(ch, end=" ")
````

### 2.3 Iterating over a tuple and a set

````python
for n in (10, 20, 30):
    print(n)

# Sets are unordered, so the order may vary
for item in {"a", "b", "c"}:
    print(item)
````

### 2.4 Iterating over a dictionary

| Method | Gives |
|--------|-------|
| `for k in d` / `d.keys()` | keys |
| `d.values()` | values |
| `d.items()` | (key, value) pairs |

````python
student = {"name": "Asha", "age": 21, "course": "CS"}

for key in student:
    print("key:", key)

for value in student.values():
    print("value:", value)

for key, value in student.items():
    print(f"{key} -> {value}")
````

## 3. The `range()` Function

`range()` produces an immutable sequence of integers — memory-efficient, as numbers are generated lazily.

| Form | Meaning | Example output |
|------|---------|----------------|
| `range(stop)` | 0 to stop-1 | `range(4)` → 0,1,2,3 |
| `range(start, stop)` | start to stop-1 | `range(2, 6)` → 2,3,4,5 |
| `range(start, stop, step)` | with a step | `range(0, 10, 3)` → 0,3,6,9 |

> ⚠️ The `stop` value is **never included**.

````python
print(list(range(5)))
print(list(range(2, 8)))
print(list(range(0, 20, 5)))
print(list(range(10, 0, -2)))   # counting down
````

### Looping with indexes

````python
colors = ["red", "green", "blue"]
for i in range(len(colors)):
    print(i, colors[i])
````

> 💡 Prefer `enumerate()` (Section 8) over `range(len(...))` — it is cleaner and more Pythonic.

## 4. The `while` Loop

### Syntax
```python
while condition:
    # body
```
The body runs **as long as the condition is `True`**. The condition is checked *before every iteration*.

> ⚠️ Always make sure something inside the loop eventually makes the condition `False`, otherwise you create an **infinite loop**.

````python
count = 1
while count <= 5:
    print("Count:", count)
    count += 1
````

### Example: countdown

````python
n = 5
while n > 0:
    print(n)
    n -= 1
print("Liftoff! 🚀")
````

### Example: loop with a sentinel / `while True`
A common pattern is `while True` with an explicit `break` — useful for input validation and menus.

````python
# Simulated user inputs (so the notebook runs without prompting)
inputs = iter(["abc", "-5", "42"])

while True:
    value = next(inputs)
    if value.lstrip("-").isdigit() and int(value) > 0:
        print("Accepted:", value)
        break
    print("Invalid input:", value)
````

> 🧪 In real programs replace `next(inputs)` with `input("Enter a positive number: ")`.

### `for` vs `while` — which to choose?
- Known collection or count → **`for`**
- Repeat until some event/condition → **`while`**

## 5. Loop Control: `break`, `continue`, `pass`

| Statement | Effect |
|-----------|--------|
| `break` | Exits the **nearest** loop immediately |
| `continue` | Skips the rest of the current iteration and goes to the next one |
| `pass` | Does nothing — a placeholder where code is syntactically required |

````python
# break — stop when found
for n in [3, 8, 12, 5, 20]:
    if n > 10:
        print("First number greater than 10:", n)
        break
````

````python
# continue — skip even numbers
for n in range(1, 8):
    if n % 2 == 0:
        continue
    print(n, end=" ")
````

````python
# pass — placeholder for future code
for n in range(3):
    pass   # TODO: implement later
print("\nDone")
````

## 6. The `else` Clause on Loops

Python loops can have an `else` block. It runs **only if the loop finishes normally** (i.e., was *not* terminated by `break`).

Typical use: **search** — "if not found, do something".

````python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            break            # divisor found -> not prime
    else:
        return True          # no break -> prime
    return False

print([n for n in range(1, 30) if is_prime(n)])
````

## 7. Nested Loops

A loop inside another loop. The inner loop runs **completely** for each iteration of the outer loop.

If the outer loop runs *m* times and the inner *n* times, the body executes **m × n** times.

````python
# Multiplication table
for i in range(1, 4):
    for j in range(1, 6):
        print(f"{i*j:3}", end="")
    print()
````

### Pattern printing

````python
rows = 5
for i in range(1, rows + 1):
    print("*" * i)
````

````python
rows = 4
for i in range(1, rows + 1):
    print(" " * (rows - i) + "* " * i)
````

### `break` in nested loops
`break` only exits the **innermost** loop. To exit both, use a flag, a function with `return`, or restructure the code.

````python
def find_pair(matrix, target):
    for r, row in enumerate(matrix):
        for c, val in enumerate(row):
            if val == target:
                return (r, c)    # exits both loops
    return None

grid = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print(find_pair(grid, 6))
````

## 8. Useful Helpers: `enumerate`, `zip`, `reversed`, `sorted`

### `enumerate()` — index + value

````python
langs = ["Python", "Java", "C++"]
for idx, lang in enumerate(langs, start=1):
    print(idx, lang)
````

### `zip()` — loop over several iterables in parallel
Stops at the **shortest** iterable.

````python
names  = ["Asha", "Ravi", "Meena"]
scores = [88, 92, 79]
for name, score in zip(names, scores):
    print(f"{name}: {score}")
````

### `reversed()` and `sorted()`

````python
for x in reversed([1, 2, 3]):
    print(x, end=" ")
print()
for x in sorted([5, 2, 9, 1], reverse=True):
    print(x, end=" ")
````

## 9. Iterables and Iterators

- **Iterable** — an object you can loop over (has `__iter__`). Examples: list, str, dict.
- **Iterator** — an object that produces the next value on demand (has `__next__`).

A `for` loop internally does this:
1. Calls `iter(obj)` to get an iterator
2. Repeatedly calls `next(iterator)`
3. Stops when `StopIteration` is raised

````python
it = iter([10, 20, 30])
print(next(it))
print(next(it))
print(next(it))
try:
    next(it)
except StopIteration:
    print("No more items")
````

### Custom iterator with a generator (`yield`)

````python
def countdown(n):
    while n > 0:
        yield n
        n -= 1

for x in countdown(5):
    print(x, end=" ")
````

## 10. Comprehensions and Generator Expressions

Concise, readable ways to build collections from loops.

| Type | Syntax |
|------|--------|
| List | `[expr for x in it if cond]` |
| Set | `{expr for x in it}` |
| Dict | `{k: v for x in it}` |
| Generator | `(expr for x in it)` |

````python
squares = [x**2 for x in range(1, 6)]
evens   = [x for x in range(10) if x % 2 == 0]
lengths = {w: len(w) for w in ["sun", "moon", "star"]}
unique  = {c for c in "banana"}

print(squares)
print(evens)
print(lengths)
print(unique)
````

````python
# Generator expression: lazy, memory-efficient
total = sum(x**2 for x in range(1_000_000))
print(total)
````

> 💡 Use comprehensions for simple transformations/filters. If the logic needs multiple statements or becomes hard to read, use a regular loop.

## 11. Common Loop Patterns

````python
# 1. Accumulator (sum)
total = 0
for n in [4, 7, 1, 9]:
    total += n
print("Sum:", total)

# 2. Counter
text = "mississippi"
count = {}
for ch in text:
    count[ch] = count.get(ch, 0) + 1
print("Frequency:", count)

# 3. Finding maximum manually
nums = [14, 3, 99, 27]
largest = nums[0]
for n in nums[1:]:
    if n > largest:
        largest = n
print("Max:", largest)

# 4. Filtering into a new list
passed = []
for score in [45, 82, 67, 30, 91]:
    if score >= 50:
        passed.append(score)
print("Passed:", passed)

# 5. Flattening a nested list
nested = [[1, 2], [3, 4], [5]]
flat = [x for sub in nested for x in sub]
print("Flat:", flat)

# 6. Reading lines of a file (pattern)
# with open("data.txt") as f:
#     for line in f:
#         print(line.strip())
````

### Fibonacci with a `while` loop

````python
a, b = 0, 1
while a < 100:
    print(a, end=" ")
    a, b = b, a + b
````

## 12. Common Mistakes and Pitfalls

### ❌ 1. Infinite loop (forgetting to update the condition)
```python
i = 0
while i < 5:
    print(i)      # i never changes -> runs forever
```
✅ Fix: add `i += 1`.

### ❌ 2. Off-by-one errors
`range(1, 5)` gives 1–4, **not** 5. Use `range(1, 6)` to include 5.

### ❌ 3. Modifying a list while iterating over it

````python
nums = [1, 2, 3, 4, 5, 6]

# Wrong: skips elements
for n in nums:
    if n % 2 == 0:
        nums.remove(n)
print("Wrong result:", nums)

# Correct: build a new list
nums = [1, 2, 3, 4, 5, 6]
nums = [n for n in nums if n % 2 != 0]
print("Correct result:", nums)
````

### ❌ 4. Using `range(len(x))` unnecessarily
Use `for item in x` or `enumerate(x)` instead.

### ❌ 5. Reusing the loop variable name
The loop variable **persists** after the loop and may overwrite an existing variable.

````python
for i in range(3):
    pass
print("i after loop:", i)   # still accessible -> 2
````

## 13. Performance and Best Practices

- ✅ Iterate directly over items: `for x in items`, not by index.
- ✅ Use `enumerate`, `zip`, `dict.items()` instead of manual indexing.
- ✅ Prefer **comprehensions** and **built-ins** (`sum`, `any`, `all`, `min`, `max`) — they are faster and clearer.
- ✅ Use **generators** for large data to save memory.
- ✅ Keep loop bodies small; move logic into functions.
- ✅ Avoid deep nesting (more than 2–3 levels) — refactor.
- ✅ Move calculations that don't change **out of the loop**.
- ✅ Use `''.join(list_of_strings)` rather than `+=` for building long strings.
- ✅ Choose meaningful variable names (`student` not `x`) for readability.

````python
import time

# Comparison: loop vs comprehension vs built-in
N = 1_000_000

t = time.perf_counter()
res = []
for i in range(N):
    res.append(i * 2)
t1 = time.perf_counter() - t

t = time.perf_counter()
res = [i * 2 for i in range(N)]
t2 = time.perf_counter() - t

t = time.perf_counter()
s = sum(range(N))
t3 = time.perf_counter() - t

print(f"for-loop append : {t1:.4f}s")
print(f"comprehension   : {t2:.4f}s")
print(f"built-in sum    : {t3:.4f}s")
````

````python
# any() / all() replace many search loops
nums = [3, 8, 15, 22]
print(any(n > 20 for n in nums))   # True if at least one matches
print(all(n > 0 for n in nums))    # True if every item matches
````

## 14. Practice Exercises

Try each one yourself first, then check the solutions below.

1. Print all numbers from 1 to 50 that are divisible by 7.
2. Compute the factorial of a number using a loop.
3. Reverse a string without slicing.
4. Count vowels in a sentence.
5. Print the FizzBuzz sequence from 1 to 20.
6. Sum the digits of an integer (e.g., 4825 → 19).
7. Print a right-aligned triangle of numbers (1, 12, 123, ...) with 5 rows.

### ✅ Solutions

````python
# 1
print([n for n in range(1, 51) if n % 7 == 0])

# 2
n, fact = 6, 1
for i in range(2, n + 1):
    fact *= i
print("6! =", fact)

# 3
s, rev = "python", ""
for ch in s:
    rev = ch + rev
print(rev)

# 4
sentence = "Loops are powerful"
print(sum(1 for ch in sentence.lower() if ch in "aeiou"))

# 5
for i in range(1, 21):
    if i % 15 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)

# 6
num, total = 4825, 0
while num > 0:
    total += num % 10
    num //= 10
print("Digit sum:", total)

# 7
for i in range(1, 6):
    row = ""
    for j in range(1, i + 1):
        row += str(j)
    print(row.rjust(5))
````

## 15. Cheat Sheet

```python
# for loop
for item in iterable:           ...
for i in range(start, stop, step): ...
for i, x in enumerate(seq, start=1): ...
for a, b in zip(seq1, seq2):    ...
for k, v in d.items():          ...

# while loop
while condition:                ...
while True:
    if done: break

# control
break      # exit loop
continue   # next iteration
pass       # do nothing

# loop-else (runs if no break)
for x in seq:
    if match: break
else:
    print("not found")

# comprehensions
[f(x) for x in seq if cond]
{k: v for k, v in pairs}
{x for x in seq}
(x for x in seq)                # generator
```

| Concept | Remember |
|---------|----------|
| `range(n)` | 0 … n-1 |
| `for` | iterate over a known sequence |
| `while` | repeat until condition false |
| `break` | leaves the nearest loop |
| `else` on loop | runs only when no `break` |
| Don't | modify a list while looping over it |

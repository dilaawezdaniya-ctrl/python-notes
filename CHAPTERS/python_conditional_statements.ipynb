## 1. Introduction

Programs rarely run in a straight line. They need to **make decisions**: grant access only to valid users, apply a discount only above a certain amount, or print different messages for different inputs. **Conditional statements** let a program execute a block of code only when a condition evaluates to `True`.

Python provides three main tools for decision making:

- **`if` / `elif` / `else` statements**: run one block out of several.
- **Conditional (ternary) expressions**: choose one *value* out of two, in a single line.
- **`match` / `case` statements** (Python 3.10+): structural pattern matching.

````python
age = 20

if age >= 18:
    print("You are eligible to vote.")
````

> **Key idea:** every condition is ultimately reduced to a Boolean value, `True` or `False`. Python then decides which block of code to execute.

## 2. Comparison (Relational) Operators

Comparison operators compare two values and always return a Boolean.

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `==` | Equal to | `5 == 5` | `True` |
| `!=` | Not equal to | `5 != 3` | `True` |
| `>` | Greater than | `3 > 5` | `False` |
| `<` | Less than | `3 < 5` | `True` |
| `>=` | Greater than or equal to | `5 >= 5` | `True` |
| `<=` | Less than or equal to | `6 <= 5` | `False` |

````python
a, b = 10, 20
print("a == b :", a == b)
print("a != b :", a != b)
print("a >  b :", a > b)
print("a <  b :", a < b)
print("a >= 10:", a >= 10)
print("a <= 5 :", a <= 5)

# Strings are compared alphabetically (by Unicode code point)
print("apple" < "banana")
````

---
## 3. Logical Operators

Logical operators combine multiple conditions.

| Operator | Description | Returns True when... |
|---|---|---|
| `and` | Logical AND | both conditions are True |
| `or` | Logical OR | at least one condition is True |
| `not` | Logical NOT | the condition is False (it inverts the value) |

### Truth table

| A | B | A and B | A or B | not A |
|---|---|---|---|---|
| True | True | True | True | False |
| True | False | False | True | False |
| False | True | False | True | True |
| False | False | False | False | True |

````python
age = 25
has_id = True

print(age >= 18 and has_id)   # both True
print(age < 18 or has_id)     # one True
print(not has_id)             # inverted
````

### Short-circuit evaluation

Python evaluates logical expressions from left to right and **stops as soon as the result is known**. In `A and B`, if `A` is falsy, `B` is never evaluated. In `A or B`, if `A` is truthy, `B` is never evaluated.

````python
def check(label):
    print(f"  evaluating {label}")
    return True

print("False and ...")
result = False and check("right side")
print("  result:", result)

print("True or ...")
result = True or check("right side")
print("  result:", result)
````

> **Why it matters:** short-circuiting lets you write safe checks such as `if items and items[0] > 5:` without raising an `IndexError` on an empty list.

## 4. The if Statement

The `if` statement executes a block of code only when its condition is true. The block is defined by **indentation** (4 spaces is the standard) and the condition line must end with a colon.

```python
if condition:
    statement_1
    statement_2
```

````python
temperature = 35

if temperature > 30:
    print("It's a hot day.")
    print("Stay hydrated!")

print("This line always runs.")
````

> Unlike C or Java, Python uses indentation, not braces `{}`, to mark blocks. Inconsistent indentation raises an `IndentationError`.

---
## 5. The if-else Statement

Use `else` to define what happens when the condition is false. Exactly one of the two blocks runs.

```python
if condition:
    # runs when condition is True
else:
    # runs when condition is False
```

````python
number = 7

if number % 2 == 0:
    print(f"{number} is even")
else:
    print(f"{number} is odd")
````

---
## 6. The if-elif-else Ladder

When there are more than two possibilities, chain conditions with `elif` (short for *else if*). Python checks the conditions **top to bottom** and runs the first block whose condition is true; all remaining branches are skipped. The final `else` is optional and acts as the default.

```python
if condition_1:
    ...
elif condition_2:
    ...
elif condition_3:
    ...
else:
    ...
```

````python
marks = 72

if marks >= 90:
    grade = "A"
elif marks >= 75:
    grade = "B"
elif marks >= 60:
    grade = "C"
elif marks >= 40:
    grade = "D"
else:
    grade = "F"

print(f"Marks: {marks} -> Grade: {grade}")
````

> **Order matters.** If you tested `marks >= 40` first, a score of 95 would be graded 'D'. Always place the most specific or most restrictive condition first.

---
## 7. Nested Conditions

An `if` statement can be placed inside another `if`, `elif` or `else` block. This is useful when a second decision depends on the first.

````python
username = "admin"
password = "secret123"

if username == "admin":
    if password == "secret123":
        print("Login successful")
    else:
        print("Wrong password")
else:
    print("Unknown user")
````

Deep nesting hurts readability. When possible, flatten the logic using logical operators:

````python
if username == "admin" and password == "secret123":
    print("Login successful")
else:
    print("Invalid credentials")
````

## 8. Conditional (Ternary) Expression

A conditional expression chooses between **two values** in a single line. Because it is an *expression*, it produces a value and can be used anywhere a value is allowed: assignments, function arguments, f-strings, lambdas and comprehensions.

```python
value_if_true if condition else value_if_false
```

````python
age = 17
status = "adult" if age >= 18 else "minor"
print(status)

# Equivalent if-else statement
if age >= 18:
    status = "adult"
else:
    status = "minor"
print(status)
````

### Common uses

````python
# 1. Pick the larger value
a, b = 15, 9
print("Max:", a if a > b else b)

# 2. Inside an f-string
n = 4
print(f"{n} is {'even' if n % 2 == 0 else 'odd'}")

# 3. Inside a lambda
absolute = lambda x: x if x >= 0 else -x
print(absolute(-8))

# 4. Inside a list comprehension
nums = [1, 2, 3, 4, 5, 6]
labels = ["even" if x % 2 == 0 else "odd" for x in nums]
print(labels)
````

### Chained (nested) conditional expressions

````python
score = 82
grade = "A" if score >= 90 else "B" if score >= 80 else "C"
print(grade)
````

> **Best practice:** use the ternary form for simple, readable choices. If you need more than one `else`, switch to a regular `if-elif-else` statement.

|  | if-else statement | Conditional expression |
|---|---|---|
| Type | Statement (does not produce a value) | Expression (produces a value) |
| Best for | Running blocks of code | Choosing between two values |
| Multiple lines | Yes | No (single line) |
| Usable in lambda / f-string | No | Yes |

## 9. Truthy and Falsy Values

Python does not require a condition to be a Boolean. Any object can be tested, and is considered either **truthy** or **falsy**.

The following values are **falsy**: `False`, `None`, `0`, `0.0`, `0j`, `""` (empty string), `[]`, `()`, `{}`, `set()` and `range(0)`. Almost everything else is **truthy**.

````python
values = [0, 1, -5, 0.0, "", "hi", [], [0], None, {}, True]
for v in values:
    print(f"{v!r:>6} -> {bool(v)}")
````

### Idiomatic usage

````python
items = []
if not items:                    # preferred over: if len(items) == 0
    print("The list is empty")

name = ""
display = name or "Guest"        # fallback / default value
print(display)
````

## 10. Membership, Identity and Chained Comparisons

### Membership operators: in / not in

````python
fruits = ["apple", "banana", "cherry"]
print("banana" in fruits)
print("grape" not in fruits)
print("th" in "python")
````

### Identity operator: is / is not

`is` checks whether two names refer to the **same object**, not whether they are equal. Use it for singletons such as `None`.

````python
result = None
if result is None:
    print("No result yet")
````

### Chained comparisons

Python lets you chain comparisons the way you write them in mathematics.

````python
x = 7
print(1 < x < 10)        # same as: 1 < x and x < 10
print(10 <= x <= 20)
````

## 11. match-case Statement (Python 3.10+)

`match` compares a value against a series of **patterns**. It is similar to `switch` in other languages but far more powerful: it can destructure sequences, dictionaries and objects. The wildcard `_` matches anything and serves as the default.

````python
def http_message(status):
    match status:
        case 200:
            return "OK"
        case 301 | 302:              # OR-pattern
            return "Redirect"
        case 404:
            return "Not Found"
        case code if code >= 500:    # guard condition
            return "Server Error"
        case _:
            return "Unknown"

for s in (200, 302, 404, 503, 418):
    print(s, "->", http_message(s))
````

### Matching structure

````python
def describe(point):
    match point:
        case (0, 0):
            return "Origin"
        case (x, 0):
            return f"On the x-axis at {x}"
        case (0, y):
            return f"On the y-axis at {y}"
        case (x, y):
            return f"Point at ({x}, {y})"

print(describe((0, 0)))
print(describe((5, 0)))
print(describe((2, 9)))
````

## 12. Common Pitfalls

### 1. Using = instead of ==

`=` assigns a value; `==` compares. Writing `if x = 5:` is a `SyntaxError`.

````python
x = 5
print(x == 5)
````

### 2. Confusing == with is

````python
a = [1, 2, 3]
b = [1, 2, 3]
print(a == b)    # same values?
print(a is b)    # same object in memory?
````

### 3. Comparing floats directly

Floating-point numbers are stored approximately. Use `math.isclose()` instead of `==`.

````python
import math
print(0.1 + 0.2 == 0.3)
print(math.isclose(0.1 + 0.2, 0.3))
````

### 4. The classic `or` mistake

`color == "red" or "blue"` is parsed as `(color == "red") or "blue"`, and a non-empty string is always truthy, so the condition is **always true**.

````python
color = "green"

if color == "red" or "blue":          # BUG: always True
    print("buggy check passed")

if color in ("red", "blue"):          # correct
    print("color is red or blue")
else:
    print("color is neither red nor blue")
````

### 5. Indentation and missing colon

Forgetting the colon after a condition raises a `SyntaxError`; mixing tabs and spaces or misaligning blocks raises an `IndentationError`. Configure your editor to insert 4 spaces per indent level.

## 13. Best Practices

- Prefer **positive conditions** (`if is_valid:`) over double negatives (`if not is_invalid:`).
- Use **truthiness** for emptiness checks: `if items:` instead of `if len(items) > 0:`.
- Compare with `None`, `True` and `False` using **`is`**, and values using **`==`**.
- Use **early returns / guard clauses** to avoid deep nesting.
- Replace long `or` chains with `in`: `if x in (1, 2, 3):`.
- Name Boolean variables as questions: `is_active`, `has_permission`.
- Keep ternary expressions short; never nest more than one level.
- Use `match-case` when branching on the *shape or value* of a single object.
- Always include a sensible default (`else` or `case _`) when inputs are unpredictable.

### Guard clause example

````python
def discount(price, is_member):
    if price <= 0:
        return 0                  # guard: invalid input
    if not is_member:
        return price              # guard: no discount
    return price * 0.9            # main logic, no nesting

print(discount(200, True))
print(discount(200, False))
print(discount(-5, True))
````

## 14. Practice Exercises

Try each exercise on your own first, then compare with the solutions below.

- **Leap year:** a year is a leap year if it is divisible by 4, except century years, which must be divisible by 400.
- **FizzBuzz:** for numbers 1 to 15 print `Fizz` for multiples of 3, `Buzz` for multiples of 5, `FizzBuzz` for both, otherwise the number.
- **BMI category:** compute BMI = weight / height squared and classify it as Underweight (<18.5), Normal (<25), Overweight (<30) or Obese.
- **Triangle type:** given three sides, report whether the triangle is invalid, equilateral, isosceles or scalene.

### Solution 1: Leap year

````python
def is_leap(year):
    return year % 4 == 0 and (year % 100 != 0 or year % 400 == 0)

for y in (1900, 2000, 2024, 2023):
    print(y, "->", "Leap year" if is_leap(y) else "Not a leap year")
````

### Solution 2: FizzBuzz

````python
out = []
for i in range(1, 16):
    if i % 15 == 0:
        out.append("FizzBuzz")
    elif i % 3 == 0:
        out.append("Fizz")
    elif i % 5 == 0:
        out.append("Buzz")
    else:
        out.append(str(i))
print(", ".join(out))
````

### Solution 3: BMI category

````python
def bmi_category(weight_kg, height_m):
    bmi = weight_kg / height_m ** 2
    if bmi < 18.5:
        return bmi, "Underweight"
    elif bmi < 25:
        return bmi, "Normal"
    elif bmi < 30:
        return bmi, "Overweight"
    return bmi, "Obese"

bmi, label = bmi_category(70, 1.75)
print(f"BMI = {bmi:.1f} ({label})")
````

### Solution 4: Triangle type

````python
def triangle_type(a, b, c):
    if a + b <= c or a + c <= b or b + c <= a:
        return "Not a valid triangle"
    if a == b == c:
        return "Equilateral"
    elif a == b or b == c or a == c:
        return "Isosceles"
    return "Scalene"

for sides in [(3, 3, 3), (5, 5, 8), (3, 4, 5), (1, 2, 10)]:
    print(sides, "->", triangle_type(*sides))
````

## 15. Quick Reference Cheat Sheet

| Construct | Syntax | Use when |
|---|---|---|
| if | `if cond:` | Run a block only if a condition holds |
| if-else | `if cond: ... else: ...` | Choose between two blocks |
| if-elif-else | `if c1: ... elif c2: ... else: ...` | Choose among many blocks |
| Ternary | `a if cond else b` | Choose between two values |
| Nested if | `if c1: if c2: ...` | Second test depends on first |
| match-case | `match x: case 1: ...` | Branch on value or structure (3.10+) |
| Membership | `x in seq` | Test presence in a collection |
| Identity | `x is None` | Test for the same object / None |
| Chained | `a < x < b` | Range check |

| Falsy values | Everything else |
|---|---|
| `False`, `None`, `0`, `0.0`, `""`, `[]`, `()`, `{}`, `set()` | Truthy (non-zero numbers, non-empty containers, most objects) |

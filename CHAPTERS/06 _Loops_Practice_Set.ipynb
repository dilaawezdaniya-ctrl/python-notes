### Solution 1: Sum of 1 to n

````python
def sum_1_to_n(n):
    total = 0
    for i in range(1, n + 1):
        total += i
    return total
````

### Solution 2: Count vowels

````python
def count_vowels(s):
    count = 0
    for ch in s.lower():
        if ch in "aeiou":
            count += 1
    return count
````

### Solution 3: Reverse a string

````python
def reverse_string(s):
    result = ""
    for ch in s:
        result = ch + result
    return result
````

### Solution 4: Factorial

````python
def factorial(n):
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result
````

### Solution 5: Multiples

````python
def multiples_of(k, limit):
    result = []
    for n in range(1, limit + 1):
        if n % k == 0:
            result.append(n)
    return result
````

### Solution 6: FizzBuzz

````python
def fizzbuzz(n):
    result = []
    for i in range(1, n + 1):
        if i % 15 == 0:
            result.append("FizzBuzz")
        elif i % 3 == 0:
            result.append("Fizz")
        elif i % 5 == 0:
            result.append("Buzz")
        else:
            result.append(str(i))
    return result
````

### Solution 7: Average

````python
def average(nums):
    if not nums:
        return 0
    total = 0
    for n in nums:
        total += n
    return total / len(nums)
````

### Solution 8: Count a character

````python
def count_char(s, ch):
    count = 0
    for c in s:
        if c == ch:
            count += 1
    return count
````

### Solution 9: Digit sum

````python
def digit_sum(n):
    total = 0
    while n > 0:
        total += n % 10
        n //= 10
    return total
````

### Solution 10: Palindrome check

````python
def is_palindrome(s):
    chars = [c.lower() for c in s if c.isalnum()]
    i, j = 0, len(chars) - 1
    while i < j:
        if chars[i] != chars[j]:
            return False
        i += 1
        j -= 1
    return True
````

### Solution 11: Second largest

````python
def second_largest(nums):
    largest = second = None
    for n in nums:
        if largest is None or n > largest:
            if largest is not None:
                second = largest
            largest = n
        elif n != largest and (second is None or n > second):
            second = n
    return second
````

### Solution 12: Word frequency

````python
def word_frequency(text):
    freq = {}
    for word in text.lower().split():
        word = word.strip(".,!?;:")
        if word:
            freq[word] = freq.get(word, 0) + 1
    return freq
````

### Solution 13: Primes up to n

````python
def primes_up_to(n):
    primes = []
    for num in range(2, n + 1):
        for d in range(2, int(num ** 0.5) + 1):
            if num % d == 0:
                break
        else:
            primes.append(num)
    return primes
````

### Solution 14: Fibonacci

````python
def fibonacci(n):
    result = []
    a, b = 0, 1
    for _ in range(n):
        result.append(a)
        a, b = b, a + b
    return result
````

### Solution 15: Remove duplicates (keep order)

````python
def remove_duplicates(items):
    seen = set()
    result = []
    for item in items:
        if item not in seen:
            seen.add(item)
            result.append(item)
    return result
````

### Solution 16: Multiplication table

````python
def multiplication_table(n, upto=10):
    return [f"{n} x {i} = {n * i}" for i in range(1, upto + 1)]
````

### Solution 17: Collatz steps

````python
def collatz_steps(n):
    steps = 0
    while n != 1:
        n = n // 2 if n % 2 == 0 else 3 * n + 1
        steps += 1
    return steps
````

### Solution 18: Matrix transpose

````python
def transpose(matrix):
    if not matrix:
        return []
    result = []
    for c in range(len(matrix[0])):
        new_row = []
        for r in range(len(matrix)):
            new_row.append(matrix[r][c])
        result.append(new_row)
    return result
````

### Solution 19: Pascal's triangle

````python
def pascal_triangle(rows):
    triangle = []
    for i in range(rows):
        row = [1] * (i + 1)
        for j in range(1, i):
            row[j] = triangle[i - 1][j - 1] + triangle[i - 1][j]
        triangle.append(row)
    return triangle
````

### Solution 20: Two-sum

````python
def two_sum(nums, target):
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                return (i, j)
    return None
````

### Solution 21: GCD (Euclid)

````python
def gcd(a, b):
    while b != 0:
        a, b = b, a % b
    return a
````

### Solution 22: Longest run

````python
def longest_run(s):
    best = current = 0
    prev = None
    for ch in s:
        current = current + 1 if ch == prev else 1
        best = max(best, current)
        prev = ch
    return best
````

### Solution 23: Decimal to binary

````python
def to_binary(n):
    if n == 0:
        return "0"
    bits = ""
    while n > 0:
        bits = str(n % 2) + bits
        n //= 2
    return bits
````

### Solution 24: Pyramid pattern

````python
def pyramid(n):
    rows = []
    for i in range(1, n + 1):
        rows.append(" " * (n - i) + "*" * (2 * i - 1))
    return rows
````

### Solution 28: Mini project: grade report

````python
def grade_report(scores):
    report = {}
    for name, marks in scores.items():
        avg = round(sum(marks) / len(marks), 1)
        if avg >= 90:
            grade = "A"
        elif avg >= 80:
            grade = "B"
        elif avg >= 70:
            grade = "C"
        elif avg >= 60:
            grade = "D"
        else:
            grade = "F"
        report[name] = (avg, grade)
    return report
````
